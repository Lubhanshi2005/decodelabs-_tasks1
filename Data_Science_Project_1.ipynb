```python

import pandas as pd

df = pd.read_csv('/content/student_data.csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())
```

```python

# Step 4: Data cleaning

print("Duplicate rows:", df.duplicated().sum())

# Remove duplicate rows if any exist
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nFinal grade summary:")
print(df['G3'].describe())

print("\nData is ready for analysis!")
```

```python

# Step 3: Understand the dataset

print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nStatistical summary:")
display(df.describe())
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Set the visual style
sns.set_theme(style="whitegrid")

# Graph 1: Study time vs final grade
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="studytime", y="G3", errorbar=None)
plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time (1 = Lowest, 4 = Highest)")
plt.ylabel("Average Final Grade (G3)")
plt.tight_layout()
plt.show()

# Graph 2: Absences vs final grade
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="absences", y="G3", alpha=0.7)
plt.title("Student Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")
plt.tight_layout()
plt.show()

# Graph 3: Distribution of final grades
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="G3", bins=20, kde=True)
plt.title("Distribution of Students' Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Graph 1: Study time vs final grade
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="studytime", y="G3", errorbar=None)
plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time (1 = Lowest, 4 = Highest)")
plt.ylabel("Average Final Grade")
plt.show()

# Graph 2: Absences vs final grade
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="absences", y="G3", alpha=0.7)
plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")
plt.show()

# Graph 3: Distribution of final grades
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="G3", bins=20, kde=True)
plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.show()
```

```python

# 1. Average final grade for each study-time group
print("1. Average Final Grade by Study Time")
print(df.groupby("studytime")["G3"].mean().round(2))

# 2. Average grade by number of absences
print("\n2. Absences and Final Grades")
print(df.groupby(
    pd.cut(df["absences"], bins=[-1, 0, 5, 10, 20, 100],
           labels=["0", "1-5", "6-10", "11-20", "21+"])
)["G3"].mean().round(2))

# 3. Average final grade by gender
print("\n3. Average Final Grade by Gender")
print(df.groupby("sex")["G3"].mean().round(2))

# 4. Overall average final grade
print("\n4. Overall Average Final Grade:", round(df["G3"].mean(), 2))
```

```python

# Step 7: Correlation analysis

correlations = df.corr(numeric_only=True)["G3"].sort_values(
    ascending=False
)

print("Correlation of numerical variables with final grade (G3):")
print(correlations.round(3))

# Plot the strongest relationships
plt.figure(figsize=(9, 6))
sns.barplot(
    x=correlations.drop("G3").values,
    y=correlations.drop("G3").index,
    errorbar=None
)
plt.title("Correlation of Numerical Factors with Final Grade")
plt.xlabel("Correlation with G3")
plt.ylabel("Numerical Factor")
plt.tight_layout()
plt.show()
```

```python

# Step 8: Project summary statistics

print("STUDENT PERFORMANCE ANALYSIS")
print("=" * 35)

print("\nTotal students:", len(df))
print("Total features:", len(df.columns))
print("Average final grade:", round(df["G3"].mean(), 2))
print("Highest final grade:", df["G3"].max())
print("Lowest final grade:", df["G3"].min())

print("\nStudents scoring 10 or more:",
      (df["G3"] >= 10).sum())

print("Students scoring below 10:",
      (df["G3"] < 10).sum())

print("\nAverage grade by previous period:")
print(df[["G1", "G2", "G3"]].mean().round(2))

print("\nProject summary completed!")
```

```python

# Step 10: Export project results

# Save correlation results
correlations.rename("Correlation_with_G3").to_csv(
    "student_performance_correlations.csv"
)

# Save key summary statistics
summary = pd.DataFrame({
    "Metric": [
        "Total students",
        "Total features",
        "Average final grade",
        "Highest final grade",
        "Lowest final grade",
        "Students scoring 10 or more",
        "Students scoring below 10"
    ],
    "Value": [
        len(df),
        len(df.columns),
        round(df["G3"].mean(), 2),
        df["G3"].max(),
        df["G3"].min(),
        int((df["G3"] >= 10).sum()),
        int((df["G3"] < 10).sum())
    ]
})

summary.to_csv("student_performance_summary.csv", index=False)

print("Results saved successfully!")
print("\nSummary:")
display(summary)
```

```python

# Step 11: Create a correlation heatmap

import matplotlib.pyplot as plt
import seaborn as sns

correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap - Student Performance Analysis")
plt.tight_layout()

# Save the heatmap for your project report
plt.savefig(
    "student_performance_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Heatmap created and saved successfully!")
```
